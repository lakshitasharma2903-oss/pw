Que.1 A die is rolled .What is the probability of getting :
(a) An even number .
(b) A number greater than 4.


ANS. A standard die has 6 faces : 1,2,3,4,5,6.
Total outcomes = 6
(a) Probability of getting an even number
Even number on a die = 2,4,6
Number of favourable outcomes= 3
P ( even number) =3/6=1/2

(b) Probability of getting a number greater than 4
Numbers greater than 4 = 5,6
Number of favourable outcomes=2
P ( number>4) =2/6=1/3.
